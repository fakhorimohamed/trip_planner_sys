```python
from langchain_core .tools import StructuredTool 
from pydantic import BaseModel
```

```python
class WeatherInput(BaseModel):
    city :str 
    
```

```python
def get_weather(city : str ) -> str :
    """Get the weather for a given city  

    Args:
        city (str): the name of the city to get the weather for 

    Returns:
        str: a string that describes the current weather in the given city 
    """
    return f"The current weather in {city} is sunny with a temperature of 25 degrees celsius"
```

```python
weather_tool = StructuredTool.from_function(
    func = get_weather,
    name = "get_weather",
    description = "Fetches real-time weather weather information for a given city . ",
    args_schema = WeatherInput ,    
)
```

```python
llm_with_tools = llm.bind_tools([weather_tool])
```

```python
response = llm_with_tools.invoke(" What is the weather like in New York City today? ")
```

```python
result = weather_tool.invoke({
    "city": "New York City"
})
```
